## Data contract

### 1. What does one row mean?

One row represents one client × one content item × one report date
in the `fact_content_daily_performance` table.

### 2. Which table will I use?

I will use `fact_content_daily_performance` as the primary warehouse
table for the Content Refresh lane because it contains content-level
search and analytics performance over time.

### 3. What time window will I use?

I will use March 2026 as the development month for this notebook.
I will not use the final June 2026 month to develop label logic because
it represents the natural future outcome window.

### 4. What will I predict or rank?

I will rank content items by their potential refresh opportunity.
The ideal target would be improvement in search performance after a
refresh. Where that outcome is not directly observed, I will treat
refresh opportunity as a proxy that must be validated.

### 5. What will I deliberately exclude?

I will exclude information that would only become known after the
refresh decision or that directly contains future outcome information,
because using such information would create data leakage.

## Feature explanations

### 1. gsc_impressions
Knowable at the decision moment because it comes from historical Google Search Console data collected before the refresh decision.

### 2. gsc_clicks
Knowable at the decision moment because it represents historical search traffic.

### 3. gsc_sum_position
Knowable at the decision moment because it summarizes historical search ranking.

### 4. ga4_sessions
Knowable at the decision moment because it is historical Google Analytics traffic data.

### 5. scroll_events
Knowable at the decision moment because it measures historical user engagement before the decision.

```python
features = con.sql("""
SELECT
    content_hash_id,
    gsc_impressions,
    gsc_clicks,
    gsc_sum_position,
    ga4_sessions,
    scroll_events
FROM read_parquet(
    'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/**/*.parquet'
)
WHERE month = '2026-03'
  AND gsc_data_available IS TRUE
LIMIT 10
""")

features
```

```python
import pandas as pd

df = features.df()

# Fake target (for demonstration only)
df["target"] = (df["gsc_clicks"] > df["gsc_clicks"].median()).astype(int)

# Leakage feature (direct copy of target)
df["leak_feature"] = df["target"]

df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X = df[[
    "gsc_impressions",
    "gsc_clicks",
    "gsc_sum_position",
    "ga4_sessions",
    "scroll_events",
    "leak_feature"
]]

y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42
)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

pred = model.predict(X_test)

print("Accuracy with leakage:", accuracy_score(y_test, pred))
```

```python
X = df[[
    "gsc_impressions",
    "gsc_clicks",
    "gsc_sum_position",
    "ga4_sessions",
    "scroll_events"
]]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42
)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

pred = model.predict(X_test)

print("Accuracy without leakage:", accuracy_score(y_test, pred))
```

## Leakage lesson

The `leak_feature` was created directly from the target, so it gave the
model information that would never be available at prediction time.

This caused the evaluation score to become unrealistically high.

After removing the leaking feature, the score became more realistic.

This demonstrates why future information or label-derived features must
never be included in the training data.

## Limitation

This notebook uses only the March 2026 slice of the warehouse.

The analysis is intended to demonstrate the data contract and feature
engineering process, not to build a production-quality refresh model.

Some business outcomes, such as the true improvement after a content
refresh, are not directly observed in this slice and would require
additional historical data.
