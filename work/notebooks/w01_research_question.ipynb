# 2. The Question

## Research Question

Which pages should be reviewed first for content refresh or other
content actions based on observable search and engagement signals?

## Decision

The decision is which pages should be prioritized for human review.

## Action

A content or SEO team could review the highest-ranked pages and
decide whether each page should be refreshed, expanded, protected,
pruned, or monitored.

## Cost of a Wrong Recommendation

If the recommendation is wrong, the team could spend time and
resources reviewing or changing a page that does not have a
meaningful opportunity. A more valuable page could also be missed.

## Why Data or ML Could Help

A website can contain many pages, so manually checking every page
can be difficult. Search and content-performance data can help
identify differences between pages and create a repeatable way to
prioritize them.

I will first investigate whether the data contains useful patterns
before deciding whether a machine learning model is actually needed.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

CSV_URL = "https://raw.githubusercontent.com/Nandini1313-cloud/flyrank-ml1-internship/refs/heads/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(CSV_URL)

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

df.head()
```

```python
# Basic statistics for the starter dataset

print("Total rows:", len(df))
print("Unique pages:", df["content_id"].nunique())

declining_pages = (df["trend_direction"] == "down").sum()
declining_percentage = (declining_pages / len(df)) * 100

print("Declining pages:", declining_pages)
print("Declining percentage:", round(declining_percentage, 2), "%")

print("Average impressions (90 days):", round(df["impressions_90d"].mean(), 2))
print("Average sessions (90 days):", round(df["sessions_90d"].mean(), 2))
```

### What these numbers show

The starter dataset contains a large number of content pages, which
makes reviewing every page with the same level of attention
difficult.

The share of pages with a downward trend shows that there are
observable differences in page performance that may be useful for
prioritization.

The average 90-day impressions also show that the pages have
measurable search exposure. These signals make it reasonable to
investigate whether pages can be ranked for human content review.

These numbers do not prove that a refresh will improve performance.
They only show that the dataset contains signals worth investigating.
