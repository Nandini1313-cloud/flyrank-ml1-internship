```python
from sklearn.model_selection import train_test_split

# Create an honest time-aware split
# First 80% = training data
# Last 20% = testing data

split_point = int(len(audit_df) * 0.80)

train_df = audit_df.iloc[:split_point].copy()
test_df = audit_df.iloc[split_point:].copy()

X_train = train_df.drop(columns=[TARGET_COLUMN])
y_train = train_df[TARGET_COLUMN]

X_test = test_df.drop(columns=[TARGET_COLUMN])
y_test = test_df[TARGET_COLUMN]

print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))

print("\nTraining target rows:", len(y_train))
print("Testing target rows:", len(y_test))
```

```python
print("Training index range:", train_df.index.min(), "to", train_df.index.max())
print("Testing index range:", test_df.index.min(), "to", test_df.index.max())
```

## Honest Split

I used a time-aware ordered split for this validation audit. The first 80% of observations are used for training and the final 20% are used for testing.

This avoids randomly mixing observations between training and testing and provides a more conservative validation setup.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the model
honest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train the model
honest_model.fit(X_train, y_train)

# Make predictions
y_pred = honest_model.predict(X_test)

print("Model trained successfully!")
```

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Honest Split Results")
print("--------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```
