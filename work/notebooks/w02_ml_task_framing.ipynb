## 5) Why ML beats a fixed rule here

A simple fixed rule could be:

"Refresh every content item that is more than 365 days old."

This rule is easy to understand, but it ignores other important
signals. An old page may still perform very well, while a newer page
may already be losing traffic and have a strong opportunity for
improvement.

The dataset contains multiple useful signals such as search volume,
impressions, clicks, CTR, average position, content age, time since
last update, and performance trend.

An ML-based approach can combine these signals to learn patterns in
historical data and produce a ranked list of content items with higher
refresh priority.

The output supports a real content action: the content team can
review the highest-priority pages first instead of manually checking
every page.

### Fixed rule vs ML approach

A fixed rule might use only one variable:

`content_age_days > 365 → refresh`

The ML approach would consider multiple variables together:

`content_age_days + search_volume + impressions + clicks + CTR +
average_position + trend_pct + days_since_last_update`

The goal is to rank content items by their estimated refresh opportunity.

### Content action

The final output would be a ranked list of content items.

For example:

1. Content A → High refresh priority
2. Content B → High refresh priority
3. Content C → Medium refresh priority
4. Content D → Low refresh priority

The content team can use this ranking to decide which pages to review
and refresh first.
