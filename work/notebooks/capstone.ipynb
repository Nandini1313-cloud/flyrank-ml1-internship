# Week 8 Showcase — 5-Minute Demo

## 1. The Question — 45 seconds

**Question:** Can machine learning help prioritize existing content pages for human review better than a simple rule?

The practical FlyRank content problem is that content teams may have many pages to review, but limited time to inspect every page manually. This project therefore focuses on ranking pages that appear more suitable for content-refresh review.

---

## 2. The Method — 1 minute

I used the anonymized FlyRank ML Internship dataset and created a declining-content label from the observed change in impressions.

I compared a transparent rule-based baseline with machine-learning models.

The models were evaluated using a client-holdout validation split so that the same client was not represented in both training and validation data.

The final ranking uses model scores and reason codes to help a human reviewer understand which pages should be inspected first.

---

## 3. One Chart — 1 minute

Show the **model vs baseline Precision@50 chart**.

The key point to explain:

> The Random Forest achieved a Precision@50 of 0.740, compared with 0.240 for the rule baseline on the same client-holdout validation split.

This means the model provided a stronger ranking signal for identifying pages labeled as declining in the evaluated data.

---

## 4. One Honest Result — 1 minute

The model performed better than the simple rule on the evaluated holdout.

However, this result does **not** prove that refreshing a page will improve its future performance.

The model identifies pages that are worth reviewing first. It does not predict Google's algorithm and it does not establish a causal effect from refreshing content.

---

## 5. One Recommendation — 1 minute

Start with the highest-ranked pages and review them manually.

For each page:

1. Check the reason code.
2. Check current search intent.
3. Check whether the information is outdated.
4. Check the page's content quality.
5. Decide whether a refresh is actually appropriate.

The model should support the editor's decision rather than replace it.
